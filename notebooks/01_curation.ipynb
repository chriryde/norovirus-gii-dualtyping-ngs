```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import yaml
import csv

from pathlib import Path
```

```python
NOTEBOOK_DIR: Path = Path.cwd()
PROJECT_ROOT: Path = NOTEBOOK_DIR.parent
CONFIG_PATH: Path = PROJECT_ROOT / "config" / "curation.yml"

with open(CONFIG_PATH, "r", encoding="utf-8") as file:
    config = yaml.safe_load(file)

GENOMIC_SPECIFICATION = PROJECT_ROOT / (config["paths"]["specification_csv"])
```

```python
genome_spec_df = pd.read_csv(GENOMIC_SPECIFICATION)
genome_spec_df.head(10)
```

```python
sns.set(rc={"figure.figsize":(16, 7)})
```

```python
genome_spec_df.sort_values(by="p_type", inplace=True)
sns.countplot(x="p_type", data=genome_spec_df)
plt.show()
```

```python
genome_spec_df.sort_values(by="genotype", inplace=True)
sns.countplot(x="genotype", data=genome_spec_df)
plt.show()
```
