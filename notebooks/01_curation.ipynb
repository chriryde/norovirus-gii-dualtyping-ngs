```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import yaml
import csv

from pathlib import Path
```

```python
NOTEBOOK_DIR: Path = Path.cwd()
PROJECT_ROOT: Path = NOTEBOOK_DIR.parent
CONFIG_PATH: Path = PROJECT_ROOT / "config" / "curation.yml"

with open(CONFIG_PATH, "r", encoding="utf-8") as file:
    config = yaml.safe_load(file)

COMPLETE_SPECIFICATION = PROJECT_ROOT / (config["paths"]["final_complete_full_specification"])
PARTIAL_SPECIFICATION = PROJECT_ROOT / (config["paths"]["final_partial_full_specification"])
```

```python
complete_spec_df = pd.read_csv(COMPLETE_SPECIFICATION)
complete_spec_df.head(10)
```

```python
sns.set(rc={"figure.figsize":(16, 7)})
```

```python
complete_spec_df.sort_values(by="p_type", inplace=True)
sns.countplot(x="p_type", data=complete_spec_df)
plt.show()
```

```python
complete_spec_df.sort_values(by="genotype", inplace=True)
sns.countplot(x="genotype", data=complete_spec_df)
plt.show()
```

```python
partial_spec_df = pd.read_csv(PARTIAL_SPECIFICATION)
partial_spec_df.head(10)
```

```python
sns.set(rc={"figure.figsize":(35, 12)})
```

```python
partial_spec_df.sort_values(by="p_type", inplace=True)
sns.countplot(x="p_type", data=partial_spec_df)
plt.show()
```

```python
sns.set(rc={"figure.figsize":(25, 7)})
```

```python
partial_spec_df.sort_values(by="genotype", inplace=True)
sns.countplot(x="genotype", data=partial_spec_df)
plt.show()
```
